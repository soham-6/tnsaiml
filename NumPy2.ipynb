```python
import numpy as np
```

```python
# ---- THE DATA ----

student_names = np.array(["Aisha", "Ben", "Carla", "Dev", "Eli", "Farah"])   # six students
subject_names = np.array(["Maths", "Physics", "English"])                    # three subjects

scores = np.array([[78, 65, 82],    # Aisha's three marks - one per subject, in the order above
                   [55, 71, 60],    # Ben's
                   [91, 88, 79],    # Carla's
                   [42, 38, 55],    # Dev's
                   [67, 74, 71],    # Eli's
                   [83, 97, 68]])   # Farah's

print(scores)                       # the whole table
print("shape:", scores.shape)

# Name
# Total marks
# Average
```

```python
sum = np.sum(scores, axis=1) 
mean = np.round(np.mean(scores, axis=1), 1)
for i in range(len(student_names)):
    print(f"Name: {student_names[i]}, Total marks:{sum[i]}, Average:{mean[i]}")
```

```python
# Subject average
subavg = np.mean(scores, axis=0)
print(subavg)
print(f"Hardest Subject: {subject_names[subavg.argmin()]}")
print(f"Easiest Subject: {subject_names[subavg.argmax()]}")
```

```python
# Top of the class
print(f"Top of the class is: {student_names[mean.argmax()]}")
```

```python
low_marks = scores<50
at_risk = low_marks.any(axis=1)
print(f"At risk student is: {student_names[at_risk]}")
```
