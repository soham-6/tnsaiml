```python
import numpy as np
```

```python
marks = np.array([35, 68, 75, 86, 29, 95])
print(marks)
```

```python
print(type(marks))
```

```python
marks2 = np.reshape(marks, (2, 3))
print(marks2)
```

```python
mean = np.mean(marks2)
print(mean)
```

```python
mean = np.mean(marks2, axis=0)     #columns mean
print(mean)
```

```python
mean = np.mean(marks2, axis=1)     #rows mean
print(mean)
```

```python
transpose = marks2.T
print(transpose)
```

```python
marks = np.array([35, 68, 75, 86, 29, 95])
print(marks)
print(marks[(marks<40)|(marks>80)])
```
