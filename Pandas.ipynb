```python
import pandas as pd
```

```python
# Create a Series called temps. Use days of the week as index.
# Print temperature of Wed.
# Print average of all days.

temps = pd.Series([18, 22, 31, 27], index=["Mon", "Tue", "Wed", "Thu"])
print(temps)
```

```python
temps_wed = temps["Wed"]
print(temps_wed)
```

```python
temps_mean = temps.mean()
print(temps_mean)
```
